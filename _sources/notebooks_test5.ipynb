# 結論與展望

&nbsp;&nbsp;&nbsp;&nbsp;本次試著去結合 GIS 與互動地圖，展示了台南安平港的主要釣點資訊。通過 iOcean 垂釣回報平台數據的處理與分群分析，對比了 ArcGIS 和 QGIS 的性能，結果顯示兩款工具在分析結果上無明顯差異，但 QGIS 在成本效益、處理速度及擴展性方面表現優異，更適合廣泛應用。此外，基於 Python 和 JavaScript 的交互式地圖設計，最終採用原生 Leaflet.js 成功解決靜態頁面互動功能失效的問題，實現了釣點展示、按鈕切換及彈跳視窗功能的穩定運行。

&nbsp;&nbsp;&nbsp;&nbsp;雖然達成了交互式地圖設計與實現的目標，但仍有許多值得探索的延伸方向。未來可以進一步針對漁獲數據進行視覺化分析，例如利用動態圖表展示釣點的漁獲種類分布、數量變化及季節性趨勢，幫助釣客選擇最佳釣魚時機與地點。同時，結合即時環境資訊如天氣預報、潮汐水位和風速等數據，可提升釣魚行程規劃的安全性與成功率。此外，整合簡易推薦系統，基於使用者偏好與環境條件提供個性化釣點與釣法建議，也能進一步強化使用體驗。
