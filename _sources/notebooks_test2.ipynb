### 校正安平漁港範圍


1. 計算平移差值：
   - 根據參考點 (`reference_coord`) 和目標點 (`target_coord`) 計算經度和緯度的平移差值。這邊的參考點以及目標點是在QGIS中操作得出。

2. 平移座標：
   - 對 `wgs84_coordinates1` 中的所有點進行平移，計算平移後的經緯度座標，結果存儲在 `shifted_coordinates` 中。

3. 顯示地圖：
   - 最後，顯示生成的地圖，地圖上包含了安平漁港和安平商港的範圍以及標記點。

```python
import folium

# 初始化地圖，設定中心點和縮放級別
m = folium.Map(location=center, zoom_start=14, width='100%', height='100%')

# 定義原始經緯度座標
# 安平漁港轉換過後之wgs84坐標
wgs84_coordinates1 = convert_coordinates(FishingPort1_twd97_coordinates)

# 目標點和參考點
reference_coord = (22.984934837581722, 120.13615556442761)
target_coord = (22.983253741376927, 120.1442680578393)

# 計算經度和緯度的平移差值
lat_diff = target_coord[0] - reference_coord[0]
lon_diff = target_coord[1] - reference_coord[1]

# 平移所有點位
shifted_coordinates = [(lat + lat_diff, lon + lon_diff) for lat, lon in wgs84_coordinates1]

# 創建一個多線段來連接所有點
folium.Polygon(locations=shifted_coordinates, color="red", fill=True, fill_color="red", fill_opacity=0.4).add_to(m) # 平移後的安平漁港
folium.Polygon(locations=wgs84_coordinates2, color="red", fill=True, fill_color="red", fill_opacity=0.4).add_to(m) # 安平商港
# 創建並添加每個點的標記和彈出視窗
for coord, orig_coord in zip(shifted_coordinates, FishingPort1_twd97_coordinates):
    popup_text = f"TWD97: {orig_coord}"
    folium.Marker(location=coord, popup=popup_text).add_to(m)

# 同步顯示安平商港的範圍
for (x, y), coord in zip(FishingPort2_twd97_coordinates, wgs84_coordinates2):
    popup_text = f"TWD97: ({x}, {y})"
    folium.Marker(location=coord, popup=popup_text).add_to(m)
# 顯示地圖
m
```
